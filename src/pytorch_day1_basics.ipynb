```python
import torch

# Detect GPU (MPS for Apple Silicon)
device = "mps" if torch.backends.mps.is_available() else "cpu"
print("Using device:", device)
```

```python
# -----------------------
# Section 1: Tensor Creation
# -----------------------
A = torch.tensor(7, device=device)
B = torch.tensor([3, 1, 4, 1, 5], device=device)
C = torch.tensor([[1, 2, 3], [4, 5, 6]], device=device)
D = torch.eye(3, device=device)

print("\nTensor A:", A)
print("\nTensor B:", B)
print("\nTensor C:", C)
print("\nTensor D:", D)

E = torch.rand(2,5, device=device)
F = torch.linspace(0, 1, steps=10, device=device)

print("\nTensor E (random):", E)
print("\nTensor F (linspace):", F)

G = torch.zeros((4,4), device=device)
H = torch.ones((4,4), device=device)
I = torch.full((4,4), 5, device=device)

print("\nTensor G (zeros):", G)
print("\nTensor H (ones):", H)
print("\nTensor I (full of 5s):", I)
```

```python
# -----------------------
# Section 2: Basic operations
# -----------------------

a = torch.tensor([2, 4, 6])
b = torch.tensor([1, 3, 5])
print("\nTensor (a + b):", a + b)
print("Tensor (a * b):", a * b)
print("Tensor F (a / b):", a / b)

d = torch.tensor([[1, 2], [3, 4]])
e = torch.tensor([[5, 6]])
print("\nTensor (d * e):", d * e)
print("Tensor (e * d):", e * d)

f = torch.rand(2,2,2,2,2)
g = torch.rand(2,2,2,2,2)
print("\nTensor (f dot g):", torch.sum(f * g))
print("Tensor L2 norm:", torch.norm(f, p=2))
```

```python
# -----------------------
# Task 3: Reshaping
# -----------------------
G = torch.arange(0, 12, device=device)
print("\nTensor G:", G)

print("Reshaped G (3,4):\n", G.reshape(3, 4))
print("Reshaped G (4,3):\n", G.reshape(4, 3))
print("Flattened G:", G.flatten())
```

```python
# -----------------------
# Section 3: Reshaping & Indexing
# -----------------------

H = torch.arange(12, device=device).reshape(3, 4)
print("\nTensor H:", H)
print("\nFirst row of Tensor H:", H[0,:])
print("\nLast column of Tensor H:", H[:,-1])
print("\nA sub matrix of H:", H[0:2,0:2])
```

```python
# -----------------------
# Optional Task : test GPU
# -----------------------
if device == "mps":
    X = torch.ones((2,2), device=device)
    Y = torch.ones((2,2), device=device)
    print("\nGPU Matrix Multiply (MPS):", X @ Y)
else:
    print("\nGPU (MPS) not available.")
```
