```python
import torch

# Detect GPU (MPS for Apple Silicon)
device = "mps" if torch.backends.mps.is_available() else "cpu"
print("Using device:", device)
```

```python
# -----------------------
# Section 1: Broadcasting
# -----------------------

v = torch.rand(3,)
M = torch.rand(3, 4)
print("v + M:", v.reshape(3,1) + M)

c = torch.rand(3,1)
r = torch.rand(1,4)
print("\n\nc + r:", c * r)

d = torch.full((5,5), 1.0)
s = torch.rand(1,5)
print("\n\nd:", d)
print("s:", s)
print("d scaled by s", d * s)
```

```python
# -----------------------
# SECTION 2 — Reductions (mean, sum, norms)
# -----------------------

A = torch.arange(1., 13., device=device).reshape(3, 4)
print("\nTensor A:\n", A)
print("Row wise Mean:\n", torch.mean(A, dim=1))
print("Column wise Mean:\n", torch.mean(A, dim=0))
print("Sum of all elements in A:", torch.sum(A))
print("Row wiseL2 Norm of A:", torch.norm(A, p=2, dim=1))

B = torch.arange(1., 51., device=device).reshape(5, 10)
print("\n\n5 vectors of 10 length:\n", B)
print("max of each vector:", B.max(dim=1).values)
print("Index of Max of each vector:", B.argmax(dim=1))
print("Vector with greatest sum:", B[B.sum(dim=1).argmax()])
```

```python
# -----------------------
# SECTION 3 — Indexing + Masking
# -----------------------

A = torch.arange(1., 26., device=device).reshape(5, 5)
print("Tensor A:", A)
print("Diagonal of A:", A.diag().diag())
print("Diagonal of A as 0:", A.fill_diagonal_(0))
print("Elements greater than A.mean():", A[A > A.mean()])

B = torch.randint(1,101,(10,10))
print("\n\nTensor B:\n", B)
print("Mask for even numbers in B:\n", B[B%2 == 0])
B_zeroed = B.clone()
B_zeroed[B < 50] = 0
print("Mask for numbers less than 0.5 in B:\n", B_zeroed)
```

```python
# -----------------------
# SECTION 4 — Autograd Basics
# -----------------------

x = torch.tensor(3.0, requires_grad=True, device=device)
y = x**3 + 2*x
y.backward()
print("Gradient of function y is :", x.grad)

x = torch.tensor([1.0,2.0,3.0], requires_grad=True, device=device)
y = (x**2).sum()
y.backward()
print("\n\nGradient of function y is :", x.grad)

x = torch.tensor([2.0], requires_grad=True, device=device)
y = 5*(x**2) + 4*x + 1
y.backward()
print("\n\nGradient of function y is :", x.grad)

x = torch.tensor(2.0, requires_grad=True)
y = x**2
z = 3*y + 4
z.backward()
print("\n\nGradient of function z is :", x.grad)
```
