```python
import torch

# Detect GPU (MPS for Apple Silicon)
device = "mps" if torch.backends.mps.is_available() else "cpu"
print("Using device:", device)
```

🔹 Section 1 — Matrix Multiplication (torch.matmul)

```python
x = torch.randn(5, device=device)
A = torch.randn(5, 5, device=device)

print(x@x)
print(A@x)
print(A@A)

print(torch.matmul(x, x))
print(torch.matmul(A, x))
print(torch.matmul(A, A))
```

🔹 Section 2 — Broadcasting in Matrix Multiplication

```python
batch_vectors = torch.randn(10, 5, device=device)
v = torch.randn(5, device=device)


print("shape of batch_vectors:", batch_vectors.shape)
print("shape of v:", v.shape)
print("shape of batch_vectors @ v: ", torch.matmul(batch_vectors, v).shape)
print("shape of v @ batch_vectors.T: ", torch.matmul(v, batch_vectors.T).shape)
```

🔹 Section 3 — einsum: Explicit Control of Math

```python
# Examples of einsum

a = torch.randint(1,5,(5,), device=device)
b = torch.randint(1,5,(5,), device=device)
c = torch.randint(1,5,(5,5), device=device)
d = torch.randint(1,5,(5,5), device=device)

print("A:", a)
print("B:", b)
print("C:", c)
print("D:", d)

# Dot product
print("Vector dot Product:", torch.einsum('i,i', a,b))

# Element wise product
print("Vector element wise Product:", torch.einsum('i,i->i',a,b))

# Outer product
print("Outer product:", torch.einsum('i,j->ij',a,b))

# Matrix vector multiplication

print("Matrix vector product:", torch.einsum('ij,j->i',c,b))

# Row wise dot product
print("Row wise dot product:", torch.einsum('ij,ij->i',c,d))

# Batch matrix vector multiplication
A = torch.randint(1,5,(10,5,5), device=device)
v = torch.randint(1,5,(10,5), device=device)

print("Batch matrix A:", A)
print("Batch vector v:", v)

print("Batch matrix vector product:", torch.einsum('ijk, ik->ij',A,v))

# Batch matrix matrix multiplication
B = torch.randint(1,5,(10,5,5), device=device)
print("Batch matrix B:", B)

print("Batch matrix matrix product:", torch.einsum('ijk,ikl -> ijl',A,B))

# Columns wise sum of matrix

print("Column wise sum of matrix:", torch.einsum('ij -> j',c))

# Weighted sum of features

X = torch.randint(1,5,(10,5), device=device)
w = torch.randint(1,5,(5,), device=device)
print("Feature matrix X:", X)
print("Weight vector w:", w)
print("Weighted sum of features:", torch.einsum('ij,j ->i',X,w))

# Attention score matrix

Q = torch.randint(1,5,(10,5,5), device=device)
K = torch.randint(1,5,(10,5,5), device=device)
print("Query matrix Q:", Q)
print("Key matrix K:", K)
print("Attention score matrix:", torch.einsum('ijk, ikj -> ijk',Q,K))
```

 Section 4 — Linear Algebra with torch.linalg

```python
M = torch.randint(1,5,(3,3), dtype=torch.float32)

# Eigenvalues and eigenvectors
eigenvalues, eigenvectors = torch.linalg.eig(M)
print("Matrix M:", M)
print("\nEigenvalues:", eigenvalues)
print("\nEigenvectors:", eigenvectors)

# SVD

U, S, Vh = torch.linalg.svd(M)
print("\nLeft singular vectors U:", U)
print("\nSingular values S:", S)
print("\nRight singular vectors Vh:", Vh)

#reconstruct M from SVD
S_matrix = torch.zeros_like(M)
S_matrix[0:S.shape[0], 0:S.shape[0]] = torch.diag(S)
M_reconstructed = U @ S_matrix @ Vh
print("\nReconstructed M from SVD:", M_reconstructed)

# Matrix Norms
frobenius_norm = torch.linalg.norm(M, ord='fro')
print("\nFrobenius norm of M:", frobenius_norm)
```

🔹 Section 5 — Manual Gradient Descent (Scalar Case)

```python
x = torch.tensor(3.0, requires_grad=True, device=device)

for step in range(10):
    y = x**3 - 4*x
    y.backward()
    with torch.no_grad():
        x -= 0.1 * x.grad
        x.grad.zero_()
    print(f"Step {step+1}: x = {x.item()}, y = {y.item()}")

# print("Value of y:", y)
# print("Gradient dy/dx at x=3:", x.grad)
```
